# 02 — Multi-Head Attention

Multi-head attention runs **H parallel attention heads** on different projection subspaces, then concatenates and projects back to `d_model`.

```python
import sys
sys.path.insert(0, '..')

import torch
import matplotlib.pyplot as plt

from models.attention import MultiHeadAttention
from visualizations.attention_heatmaps import plot_multi_head_attention

d_model, n_heads = 128, 4
batch, seq_len = 2, 6

mha = MultiHeadAttention(d_model, n_heads, dropout=0.0)
x = torch.randn(batch, seq_len, d_model)

out, weights = mha(x, x, x)
print('Input: ', x.shape)
print('Output:', out.shape)
print('Weights:', weights.shape)  # (batch, n_heads, seq_len, seq_len)
```

```python
fig = plot_multi_head_attention(weights[0].detach().numpy(), n_heads)
plt.suptitle('Multi-Head Self-Attention (batch 0)')
plt.show()
```
